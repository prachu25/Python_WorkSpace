```python
# Updating film collections

# condition
from os import name
fnm = input('Enter a film Name: ')
condition = {}
condition['filmnm'] = fnm
print(condition)
```

```python
# change
rate = int(input('Enter new imdbrating: '))
change = {}
change['imdbrating'] = rate
print(change)

upd = {}
upd['$set'] = change
print(upd)
```

```python
cnt = coll_film.update_one(condition,upd)
print(cnt.modified_count)

if cnt.modified_count > 0:
  print('Rating changed successfully')
else:
  print('film name not found')
```

```python
db2 = client['shoppingdb']
```

```python
coll_users = db2['users']
```

```python
uid = input('Enter user_id: ')
psw = input('Enter a Password: ')

condition = {}
condition['userid'] = uid

change = {}
change['password'] = psw

# print(condition)
# print(change)

upd = {}
upd['$set'] = change
# print(upd)

result = coll_users.update_one(condition,upd)
print()

if result.modified_count > 0:
  print('Password changed successfully')
else:
  print('user not found!')
```
