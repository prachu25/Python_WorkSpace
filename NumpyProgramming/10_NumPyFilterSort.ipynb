```python
import numpy
```

**Filter**

```python
arr = numpy.random.randint(1,75,10)
print(arr)
```

```python
filtered = arr[arr>50]
print(filtered)   # stering and then printing
```

```python
print(arr[3:8])    # direct without stored
```

```python
# direct without storing
print(arr[arr%2 == 0])
print(arr[arr%2 == 1])
```

**# SOME OPERATIIONS**

```python
arr = numpy.array(['Prachi','Pooja','Sujal','akash','chiku','anay'])
print(arr)
```

**1.startswith()**

```python
print(arr[numpy.char.startswith(arr,'P')])
```

**2.endswith()**

```python
print(arr[numpy.char.endswith(arr,"a")])
```

**3.upper()**

```python
print(numpy.char.upper(arr))
```

**4.lower()**

```python
print(numpy.char.lower(arr))
```

**5.replace()**

```python
print(numpy.char.replace(arr,"a","@"))
```

**6.find()**

```python
print(numpy.char.find(arr,"a"))   # -1 means character doesn't found
```

**7.count()**

```python
print(numpy.char.count(arr,"a"))
```

**8.split()**

```python
a = numpy.array(["hello world", "numpy python"])

print(numpy.char.split(a))
```

**9.join()**

```python
print(numpy.char.join("-",arr))
```

```python
print(numpy.char.join("+",arr))
```

**10.sorting()**

```python
arr1 = numpy.array(['Mohit','Ruru','Aakansha','yogesh','gulu','Anushka','Rahul'])
print(arr1)
print(numpy.sort(arr1))
```

**11. contains filtering**

```python
print(arr1[numpy.char.find(arr1,"a")>=0])
```

**12. string length**

```python
print(numpy.char.str_len(arr1))
```

**13. capitalize()**

```python
print(numpy.char.capitalize(arr1))
```

```python
a = numpy.array(['hello world'])
print(numpy.char.capitalize(a))   # only the first letter

print()
print(numpy.char.title(a))   # every
```

**Sort**

```python
arr = numpy.random.randint(1,20,10)
print(arr)  # NOTE - numpy allow a duplicate
```

```python
print(numpy.sort(arr))
```

```python
print(numpy.sort(arr)[::-1])   # reverse desc
```

**Unique**

```python
arr = numpy.array([9,24,13,45,45,9,13,44,68])
print(arr)    # numpy easily stored a Duplicate value
```

```python
# UNIQUE
un = numpy.unique(arr)
print(un)
```
