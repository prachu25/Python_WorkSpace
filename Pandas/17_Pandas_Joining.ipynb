**In Pandas, joining means combining data from two DataFrames into one DataFrame.**

**It is similar to SQL joins.**

```python
import pandas
```

**Doctor URL**

```python
docurl = "http://raw.githubusercontent.com/sohamglobal/datasets/refs/heads/main/doctors.csv"
dfdoc=pandas.read_csv(docurl)
print(dfdoc)
```

**Patient URL**

```python
paturl = "http://raw.githubusercontent.com/sohamglobal/datasets/refs/heads/main/patients.csv"
dfpat=pandas.read_csv(paturl)
print(dfpat.head())
```

**# Inner join between patient and doctor tables using doctor_id**

new_df = dataframe1.merge(dataframe2, on="common_column", how="join_type")

print(new_df)

```python
res = dfpat.merge(dfdoc,on="doctor_id", how="inner")
print(res)
```

```python
res = dfpat.merge(dfdoc,on="doctor_id", how="outer")
print(res)
```

```python
res = dfpat.merge(dfdoc,on="doctor_id", how="right")
print(res)
```
