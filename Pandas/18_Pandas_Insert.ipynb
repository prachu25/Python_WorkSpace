**In Pandas, insert means adding a new column at a specific position in a DataFrame.**

**we use insert()**

```python
import pandas
```

```python
# first create columns to add data
df=pandas.DataFrame(columns=['Name','Science','English','Math'])
print(df)
```

```python
# insert at index 0
df.loc[0]=['Owi',88,96,99]
print(df)
```

```python
# insert at index 1
df.loc[1]=['Rishi',68,96,89]
print(df)

```

```python
# loc[] overwrites existing row data if index already exists
df.loc[0]=['Joe',78,99,77]
print(df)
```

```python
# loc[] assigns values only to specified columns of a particular row
df.loc[2,['Name','Math']]=['Yogesh',99]
print(df)
```

```python
# we can directly give a one value
df.loc[2,'Science']=88
```

```python
print(df)
```

```python
dic={
    'country':['usa','england','france','germany','japan'],
    'capital':['washington','london','paris','berlin','tokyo']
}

df=pandas.DataFrame(dic)
print(df)
```
