```python
import pandas as pd
import matplotlib.pyplot as plt
# Import the Seaborn library for creating beautiful and advanced data visualizations
import seaborn as sns
```

```python
data={
    "Product":["Mobile","Laptop","Mobile","Laptop","Mobile","Laptop","Mobile","Laptop"],
    "Brand":["Samsung","Dell","Apple","HP","OnePlus","Lenovo","Xiaomi","Asus"],
    "PayMode":["UPI","Card","Card","UPI","UPI","UPI","COD","Card"],
    "Amount":[25000,55000,70000,60000,32000,51000,20000,65000]
}
df= pd.DataFrame(data)
print(df)
```

```python
df= pd.read_csv("sales_data.csv")
print(df.head())
```

```python
# Group data by Product and calculate total sales amount for each product
prodsales = df.groupby("Product")["Amount"].sum()
print(prodsales)
```

```python
plt.figure(figsize=(16,8))
plt.bar(prodsales.index,prodsales.values)    #index-x-axis value-y-axis
plt.title("Total sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales Amount")
plt.show()
```

**value_counts() is used to count the frequency of unique values in a column.**

```python
# Count how many times each payment mode is used
paycounts = df['PayMode'].value_counts()
print(paycounts)
```

```python
plt.figure(figsize=(6,3))
plt.bar(paycounts.index,paycounts.values)

plt.xlabel("Payment Mode")
plt.ylabel("Count")
plt.title("Payment Mode Usage")

plt.show()
```

```python
# create Pie chart
plt.figure(figsize=(6,6))
plt.pie(
    paycounts.values,
    labels=paycounts.index,
    autopct='%1.1f%%'
)

plt.title("Payment Mode Distribution")
plt.show()
```

```python
# Create a figure with width 7 and height 4
plt.figure(figsize=(7,4))

# Create histogram of Amount column
plt.hist(df['Amount'], bins=5)

# Add chart title
plt.title("Sales Amount Distribution")

# Label for X-axis
plt.xlabel("Amount")

# Label for Y-axis
plt.ylabel("Frequency")

# Display the chart
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Product',data=df)
plt.title("Count of Products sold")
plt.show()
```
