```python
import pandas
```

```python
url = "http://sohamspring.azurewebsites.net/api/accounts/all"

df = pandas.read_json(url)
print(df.head())
```

```python
print(df.shape)
```

```python
print(df['accnm'].head(3))
```

```python
print(df[['accnm','balance']].head())   # head-> starting records
```

```python
print(df[['accnm','acctype']].tail())  # tail-> last records
```

```python
# Select row  3
print(df.loc[3])
```

```python
print(df.loc[5:9])    # from 5th row to 9th col
```

```python
# Select accnm value from index 13
print(df.loc[13, 'accnm'])
```

```python
# Select multiple columns from index 13
print(df.loc[13, ['accnm', 'acctype', 'balance']])
```

```python
print(df.loc[31:35,'accnm'])
```

```python
print(df.loc[31:35,['accnm','balance','acctype']])
```
