```python
import pandas
```

```python
df = pandas.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],

    "dob": [
        "1990-05-15",
        "1985-11-23",
        "2000-07-04",
        "1978-03-30",
        "1995-12-01"
    ],

    "join_date": [
        "2020-01-10",
        "2019-06-15",
        "2021-08-20",
        "2018-03-05",
        "2022-11-30"
    ],

    "last_login": [
        "2024-03-01 08:30:00",
        "2024-03-05 14:15:00",
        "2024-02-28 09:00:00",
        "2024-03-07 17:45:00",
        "2024-03-06 11:20:00"
    ]
})

print(df)
```

```python
df['dob']=pandas.to_datetime(df['dob'])   # dob string ko -> date convert

df['dob'] = pandas.to_datetime(df['dob']).dt.strftime('%d %B %Y')
print(df)
```

```python
df['join_date']=pandas.to_datetime(df['join_date'])
df['last_login']=pandas.to_datetime(df['last_login'])
```
