```python
from pymongo import MongoClient     # Imports MongoClient to connect Python with MongoDB.
```

**Creates a connection to the MongoDB database cluster using the given connection string.**

```python
client = MongoClient(["CONNECTION_STRING_PYTHON_SHELL"])
```

```python
db = client["spiderdb"]
```

```python
coll=db["students"]
```

```python
for doc in coll.find():
  print(doc)
```

```python
collfilm = db["films"]
```

```python
for document in collfilm.find():
  print(document)
```

```python
for doc in collfilm.find():
  print(f"Film Name: {doc['filmnm']} | Genere: {doc['genre']} | Imdbrating: {doc['imdbrating']} ")
```

```python
for doc in collfilm.find():
  try:
      print(f"Film Name: {doc['filmnm']} | Genere: {doc['genre']} | Imdbrating: {doc['imdbrating']} | Actor: {doc['actor']}")
  except:
    print('not found')
```
