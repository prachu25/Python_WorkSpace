**Retriving Data Conditionally**

---
**# Filtering means selecting specific rows based on conditions**


```python
import pandas
```

```python
url = "http://sohamspring.azurewebsites.net/api/accounts/all"
df=pandas.read_json(url)
print(df.head())
```

```python
print(df[df['acctype']=='ppf'])
```

```python
print(df[df['acctype']=='saving'])
```

```python
# print the acctypes which is not saving
print(df[df['acctype']!='saving'])
```

```python
print(df['acctype'])
```

```python
print(df['acctype'].unique())  # Unique acctype
```

```python
# print(df[ () & () ])
print(df[(df['acctype']=='ppf') & (df['balance']>75000)])
```

**AND &**

```python
print(df[(df['acctype']=='fixed') & (df['balance']<50000)])
```

**OR |**

```python
print(df[(df['acctype']=='fixed') | (df['balance']<50000)])
```
