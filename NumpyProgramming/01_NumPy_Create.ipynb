**What is NumPy?**

 NumPy is a Python library used for fast numerical operations using arrays.

 everything we do in data science needs numpy.




**Uses:**

 Math calculations

 Working with large data

 Data Science & Machine Learning



**Why better than list?**

✔ Faster

✔ Uses less memory

✔ Easy calculations (no loops needed)

✔ Supports multi-dimensional arrays

**Linear spacing** in Python means creating numbers with equal gap (spacing) between them.

Mostly it is done using NumPy linspace() function.

**numpy.linspace(start, stop, number_of_values)**

```python
x = numpy.linspace(1,10,5)
print(x)

# Explanation:

# Start = 1
# Stop = 10
# Total values = 5
# All numbers have equal spacing.
```

```python
q = numpy.linspace(1,13,9)
print(q)
```

**Multiple types**

```python
lst = ['Rohit',45,2.62,True]

for i in lst:
  print(type(i))
```

```python
arr = numpy.array(lst)
print(arr)                  # numpy array stored a homogenous data type

# All elements in NumPy array are of same data type.
# different types are given, NumPy converts them into one common type.
```
