**In Pandas, grouping means combining rows that have the same value and then performing operations like sum(), count(), mean(), etc.**

```python
import pandas
```

```python
url="http://sohamspring.azurewebsites.net/api/accounts/all"
df=pandas.read_json(url)
print(df.head())
```

```python
dfgroups=df.groupby('acctype')   # data stored inside the dfgroups-> i.e object
print(dfgroups)
```

**# groupby() creates a GroupBy object, actual data appears after applying operations**

```python
for group in dfgroups:
  print(group)
```

```python
print(df['acctype'].value_counts())
```
