```python
import numpy
import pymysql
```

```python
connection = pymysql.connect(
    host=('HOST_NAME'),
    port=('PORT_NUMBER'),
    user=('USER_NAME'),
    password=('PASSWORD'),
    database=('DATABASE_NAME')
    )
print('connected successfully')
```

```python
curs = connection.cursor()
```

```python
curs.execute("select * from customers")
data = curs.fetchall()
arr = numpy.array(data)
print(arr)
```

```python
curs.execute("select * from doctors")
d = curs.fetchmany(4)

for i in d:
  print(i)
```

**Fetching Patient and Doctor Details using INNER JOIN**

```python
curs.execute("select p.patient_name, p.disease,d.doctor_name from patients p join doctors d on d.doctor_id = p.doctor_id ;")
data = curs.fetchall()
arr = numpy.array(data)
print(arr)
```
