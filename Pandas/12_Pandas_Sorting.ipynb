```python
import pandas
```

```python
url="http://sohamspring.azurewebsites.net/api/orders/all"
```

```python
df=pandas.read_json(url)
print(df)
```

**# Sort**

```python
res=df.sort_values('company')
print(res)
```

**# Descending Sort on City**

```python
res=df.sort_values('city',ascending=False)  # asc -> Flase means---> des
print(res)
```
