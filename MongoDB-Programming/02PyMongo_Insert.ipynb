```python
collfilm = db1["films"]
```

```python
id = int(input("Enter a _id: "))
fnm = input("Enter a flim name: ")
rate = int(input("Enter a imdbrating: "))
genre = input("Enter a genre: ")
music = input("Enter a music:")
```

New Student Collection which is already presnet in db

```python
collstudent = db2['students']
```

```python
roll = int(input("Enter roll no: "))
name = input("Enter name: ")
gender = input("Enter gender: ")
city = input("Enter city: ")
email = input("Enter email: ")
mobile = input("Enter mobile: ")

skills = input("Enter skills (comma separated): ").split(",")
```

```python
dic = {}

dic['rollno'] = roll
dic['name'] = name
dic['gender'] = gender
dic['city'] = city

dic['contact'] = {
    "email": email,
    "mobile": mobile
}

dic['skills'] = skills

# check placement
placed = input("Is student placed? (yes/no): ")

if placed.lower() == "yes":
    company = input("Enter company name: ")
    role = input("Enter role/profile: ")
    package = input("Enter package: ")

    dic['placement'] = {
        "company" :  company,
        "role" :  role,
        "package" :  package
    }



print()

for key in dic:
  print(key, ":", dic[key])
```

```python
collstudent.insert_one(dic)
print("student insert successfully...")
```
