**__ add __()** is used to define how the + operator should work when adding two objects of a user-defined class.

```python
class Product:

    def __init__(self, quantity, purchcost):
        self.quantity = quantity
        self.purchcost = purchcost

    def showstock(self):
        print(f"Quantity: {self.quantity}")
        print(f"Purchase Cost: {self.purchcost}")

    def __add__(self,obj):

          # self = dell
          # obj = hp

      # q = dell.quantity + hp.quantity
      q =  self.quantity + obj.quantity
      c =  self.purchcost + obj.purchcost

      return Product(q,c)

# after we add __add__ this method the python will see internally
# total = dell.__add__(hp)

dell = Product(9,641500)
hp = Product(15,67889)

total_t = dell + hp
total_t.showstock()
```
