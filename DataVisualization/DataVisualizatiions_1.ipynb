```python
import pandas  as pd
import matplotlib.pyplot as plt
# from matplot import pyplot     # second way to import matplot library
```

**LINE Char-Growth & Decline**

```python
data = {
    "month":["Dec25","Jan26","Feb26","Mar26","Apr26","May26"],
    "sales":[20000,25000,17500,30000,21000,19700]
}

df = pd.DataFrame(data)
print(df)
```

```python
plt.plot(df["month"],df['sales'],marker = "o", color = 'green')

plt.xlabel("Month")
plt.ylabel("Sale")
plt.title("Sales in last 6 months")

plt.show()
```

**BAR Chart - Comapre Values**

```python
products = ['laptop','mobile','tablet','smartwatch']
sale = [50,110,27,65]

plt.bar(products,sale,color = 'lightgreen')
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Item sold in April 2026")
plt.show()
```

**SCATTER Chart - Identifies correlation**

```python
study_hours = [1,2,3,4,5]
marks = [40,50,60,70,90]

plt.scatter(study_hours,marks,color ='red')
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours Vs Marks")
plt.show()
```
