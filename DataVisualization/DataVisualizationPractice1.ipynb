```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Student Marks
student_marks = {
    "Rahul":77,
    "Pirya":45,
    "Joe":99,
    "kana":44,
    "Rishi":83,
    "Ishan":35,
    "Surya":58,
    "Rohit":90,
    "Virat":49
}

df = pd.DataFrame(
    student_marks.items(),
    columns=["Student" , "Marks"]
    )
print(df)
```

```python
plt.figure(figsize=(6,3))
plt.bar(df['Student'],df['Marks'])
plt.title("Student Marks Graph")
plt.xlabel("Students")
plt.ylabel("Marks")
```

```python
# Products Sales
product_sales = {
    "Laptop": 25,
    "Mobile": 40,
    "TV": 15,
    "Camera": 10,
    "Tablet": 20
}

ps = pd.DataFrame(
    product_sales.items(),
    columns=['Product','Amount'])
print(ps)
```

```python
plt.figure(figsize=(5,3))
plt.bar(ps['Product'],ps['Amount'])
plt.title("Product Distribution")
plt.xlabel("Products")
plt.ylabel("sale_Amount")
plt.show()
```

```python
plt.pie(
    ps.values,
    labels=ps.index,
    autopct='%1.1f%%'
)

plt.title("Product Wise Sales Distribution")
plt.show()
```

```python
employees = {
    "Google": 180,
    "Microsoft": 150,
    "Amazon": 380,
    "Infosys": 120,
    "TCS": 200,
    "Wipro": 140,
    "Accenture": 380,
    "IBM": 60,
    "Oracle": 130,
    "Capgemini": 170,
    "HCL": 145,
    "Cognizant": 185
}

df = pd.DataFrame(
    employees.items(),
    columns=['Company','Employees']
)
print(df)
```

plt.barh()

h = Horizontal

bar() → Vertical bars

barh() → Horizontal bars

```python
# Hoeixonta Bar Chart
plt.figure(figsize=(6,4))

plt.barh(
    df['Company'],
    df['Employees']
)

plt.title("Employees in Companies")
plt.xlabel("Employees")
plt.ylabel("Company")
plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    df['Employees'],
    labels=df['Company'],
    autopct='%1.1f%%'
)

plt.title("Company Employee Distribution")
plt.show()
```
