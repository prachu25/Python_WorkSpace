```python
import pandas
```

```python
data={
    'A':[10,20,30,40,50],
    'B':[5,15,25,35,45]
}

df=pandas.DataFrame(data)
print(df)
```

```python
print(df+3)
```

```python
# Addition
df['Add'] = df['A'] + df['B']
print(df)
```

```python
# Subtraction
df['Sub'] = df['A'] - df['B']
print(df)
```

```python
# Multiplication
df['Mul'] = df['A'] * df['B']
print(df)
```

```python
print(df['A']**2)  # power
```

```python
print(df.mean())
```
