```python
# date -> YYYY-MM-DD
arr= numpy.arange('2026-06-03', '2026-06-30', dtype='datetime64[D]')
print(arr)
```

**dtype='datetime64[D]'**

dtype → Defines the data type


datetime64 → NumPy date and time data type


[D] → D means Day


This tells NumPy to store the values as dates in day format.


Different units:

datetime64[Y] → Year

datetime64[M] → Month

datetime64[D] → Day

datetime64[h] → Hour

datetime64[m] → Minute

datetime64[s] → Second
