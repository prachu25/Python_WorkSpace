```python
data = {
    "Name": [
        "Prachi", "Rahul", None, "Amit", "Sneha",
        None, "Rohan", "Kunal", "Pooja", None
    ],

    "Age": [
        21, None, 24, 25, None,
        28, 30, None, 22, 27
    ],

    "City": [
        "Pune", "Nagpur", None, "Mumbai", "Delhi",
        "Chennai", None, "Hyderabad", None, "Bangalore"
    ],

    "Salary": [
        25000, None, 30000, 28000, None,
        35000, 40000, None, 27000, 32000
    ],

    "Department": [
        "IT", "HR", None, "Finance", "IT",
        None, "Sales", "Marketing", None, "HR"
    ]
}

df = pandas.DataFrame(data)
print(df)
```

```python
# 1. check the missing values
print(df.isnull())
```

```python
# 2. Counts the Total missing values
print(df.isnull().sum().sum())
```

```python
# 3. Counts the column wise missing values
print(df.isnull().sum())
```

```python
# 4. Fill missing value in name with 'Unknown'
df['Name'].fillna('Unknown',inplace=True)
print(df)
```

```python
# 5. Fill missing values in Age using mean
df['Age']= df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
# 6. Fill missing values in Salary using mean
df['Salary']= df['Salary'].fillna(df['Salary'].mean())
print(df)
```

```python
# 7. Removes rows having Nan
res= df.dropna()
print(res)
```

```python
# 8. Remove columns having Nan
res = df.dropna(axis=1)  # depratment colums are delete bz conatins Nan value before rows deleting operation
print(res)
```

```python
# 9. Convert Age columns into int type
df['Age']= df['Age'].astype(int)
print(df)
```
